```python
# importing dependencies

import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

Data Collection and preprocessing

```python
# loading the data from csv to a pandas datframe
raw_mail_data = pd.read_csv('/content/mail_data.csv')
```

```python
print(raw_mail_data)
```

```python
#replace the null values with a null string
mail_data = raw_mail_data.where((pd.notnull(raw_mail_data)),'')
```

```python
mail_data.head()
```

```python
mail_data.shape
```

Label Encoding

```python
# label spam mail as 0, ham mail as 1

mail_data.loc[mail_data.Category == 'spam', 'Category',] = 0
mail_data.loc[mail_data.Category == 'ham','Category',] = 1
```

spam --> 0

ham --> 1

```python
mail_data.head()
```

```python
# seperating the data as text and label

X = mail_data['Message']
Y = mail_data['Category']
```

```python
print(X)
```

```python
print(Y)
```

Splitting the datset into training and testing data

```python
X_train, X_test, Y_train, Y_test = train_test_split(X,Y,test_size = 0.2,random_state = 3)
```

```python
print(X.shape,X_train.shape,X_test.shape)
```

Feature Extraction ---> converting text data to numerical values

```python
# transform the text data to feature vectors that can be used as input to the Logistic Regression Model

feature_extraction = TfidfVectorizer(min_df =1, stop_words = 'english', lowercase= True)
X_train_features = feature_extraction.fit_transform(X_train)
X_test_features = feature_extraction.transform(X_test)

# convert Y_train and Y_test values as integers

Y_train = Y_train.astype('int')
Y_test = Y_test.astype('int')
```

```python
print(X_train)
```

```python
print(X_test_features)
```

Training the model

Logistic Regression

```python
model = LogisticRegression()
```

```python
#training the logistic regression model with the training data
model.fit(X_train_features,Y_train)
```

Model Evaluation

Evaluating the trained model

```python
prediction_on_training_data = model.predict(X_train_features)
accuracy_on_training_data = accuracy_score(Y_train,prediction_on_training_data)
```

```python
print('Accuracy on training data: ',accuracy_on_training_data)
```

```python
prediction_on_test_data = model.predict(X_test_features)
accuracy_on_test_data = accuracy_score(Y_test,prediction_on_test_data)
```

```python
print('Accuracy on training data: ',accuracy_on_test_data)
```
